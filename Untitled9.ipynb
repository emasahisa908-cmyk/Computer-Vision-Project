```python
model = YOLO('yolov8s.pt')  # pretrained YOLOv8 small

model.train(
    data="/Users/eri/Desktop/solar_panel/data.yaml",
    epochs=15,
    imgsz=640,
    batch=8,
    lr0=0.0005,
    name="solar_panel_v1"
)
```

```python
metrics = model.val()

print(f"mAP50: {metrics.box.map50:.4f}")
print(f"mAP50-95: {metrics.box.map:.4f}")
print(f"Precision: {metrics.box.mp:.4f}")
print(f"Recall: {metrics.box.mr:.4f}")

P, R = metrics.box.mp, metrics.box.mr
F1 = 2 * P * R / (P + R) if (P + R) > 0 else 0
print(f"F1-score: {F1:.4f}")
```

```python
from PIL import Image
import matplotlib.pyplot as plt

result_dir = "/Users/eri/runs/detect/solar_panel_v1"

img1 = Image.open(f"{result_dir}/results.png")
img2 = Image.open(f"{result_dir}/confusion_matrix.png")

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img1)
plt.axis('off')
plt.title('Training Results')

plt.subplot(1,2,2)
plt.imshow(img2)
plt.axis('off')
plt.title('Confusion Matrix')

plt.tight_layout()
plt.show()
```

```python
from ultralytics import YOLO

v1_weights = "/Users/eri/runs/detect/solar_panel_v1/weights/best.pt"
model_v1 = YOLO(v1_weights)

metrics_v1_test = model_v1.val(
    data="/Users/eri/Desktop/solar_panel/data.yaml",
    split="test"
)
P1, R1 = float(metrics_v1_test.box.mp), float(metrics_v1_test.box.mr)
F11 = 2*P1*R1/(P1+R1) if (P1+R1)>0 else 0.0

print("[V1 TEST] mAP50-95:", float(metrics_v1_test.box.map))
print("[V1 TEST] mAP50:", float(metrics_v1_test.box.map50))
print("[V1 TEST] Precision:", P1)
print("[V1 TEST] Recall:", R1)
print("[V1 TEST] F1:", F11)
```

```python
v2_weights = "/Users/eri/runs/detect/solar_panel_v2/weights/best.pt"
model_v2 = YOLO(v2_weights)

metrics_v2_test = model_v2.val(
    data="/Users/eri/Desktop/solar_panel/data.yaml",
    split="test"
)
P2, R2 = float(metrics_v2_test.box.mp), float(metrics_v2_test.box.mr)
F12 = 2*P2*R2/(P2+R2) if (P2+R2)>0 else 0.0

print("[V2 TEST] mAP50-95:", float(metrics_v2_test.box.map))
print("[V2 TEST] mAP50:", float(metrics_v2_test.box.map50))
print("[V2 TEST] Precision:", P2)
print("[V2 TEST] Recall:", R2)
print("[V2 TEST] F1:", F12)

# show compare table
print("\n=== TEST Comparison (V1 vs V2) ===")
print(f"{'Model':<10}{'lr0':<10}{'Epochs':<8}{'mAP50':<10}{'P':<10}{'R':<10}{'F1':<10}")
print(f"{'v1':<10}{'0.0005':<10}{'10':<8}{metrics_v1_test.box.map50:.4f}    {P1:.4f}    {R1:.4f}    {F11:.4f}")
print(f"{'v2':<10}{'0.001':<10}{'10':<8}{metrics_v2_test.box.map50:.4f}    {P2:.4f}    {R2:.4f}    {F12:.4f}")
```
