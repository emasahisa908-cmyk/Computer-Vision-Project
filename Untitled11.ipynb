```python
def show_training_results(model_name):
    path = f"/content/drive/MyDrive/solar_panel_runs/{model_name}"
    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    plt.imshow(Image.open(f"{path}/results.png"))
    plt.axis('off')
    plt.title(f"{model_name} - Training Results")

    plt.subplot(1, 2, 2)
    plt.imshow(Image.open(f"{path}/confusion_matrix.png"))
    plt.axis('off')
    plt.title(f"{model_name} - Confusion Matrix")
    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
#  Model V1 (10 epochs)
# ============================================================
model_v1 = YOLO('yolov8s.pt')
model_v1.train(
    data="/content/drive/MyDrive/solar_panel/data.yaml",
    epochs=10,
    imgsz=640,
    batch=8,
    lr0=0.0005,
    name="solar_panel_v1",
    project="/content/drive/MyDrive/solar_panel_runs"
)
```

```python
metrics_v1 = model_v1.val(data="/content/drive/MyDrive/solar_panel/data.yaml", split="test")
P1, R1 = metrics_v1.box.mp, metrics_v1.box.mr
F11 = 2 * P1 * R1 / (P1 + R1)
print(f"[V1 TEST] mAP50={metrics_v1.box.map50:.4f}, P={P1:.4f}, R={R1:.4f}, F1={F11:.4f}")
show_training_results("solar_panel_v1")
```

```python
# ============================================================
#  Model V2 (20 epochs)
# ============================================================
model_v2 = YOLO('yolov8s.pt')
model_v2.train(
    data="/content/drive/MyDrive/solar_panel/data.yaml",
    epochs=20,
    imgsz=640,
    batch=8,
    lr0=0.001,
    optimizer='adam',
    name="solar_panel_v2",
    project="/content/drive/MyDrive/solar_panel_runs"
)
```

```python
metrics_v2 = model_v2.val(data="/content/drive/MyDrive/solar_panel/data.yaml", split="test")
P2, R2 = metrics_v2.box.mp, metrics_v2.box.mr
F12 = 2 * P2 * R2 / (P2 + R2)
print(f"[V2 TEST] mAP50={metrics_v2.box.map50:.4f}, P={P2:.4f}, R={R2:.4f}, F1={F12:.4f}")
show_training_results("solar_panel_v2")
```

```python

# ============================================================
#  Model V3 (30 epochs + augmentation)
# ============================================================
model_v3 = YOLO('yolov8s.pt')
model_v3.train(
    data="/content/drive/MyDrive/solar_panel/data.yaml",
    epochs=30,
    imgsz=640,
    batch=8,
    lr0=0.001,
    optimizer='adam',
    augment=True,  # enable augmentation
    name="solar_panel_v3",
    project="/content/drive/MyDrive/solar_panel_runs"
)
```

```python
metrics_v3 = model_v3.val(data="/content/drive/MyDrive/solar_panel/data.yaml", split="test")
P3, R3 = metrics_v3.box.mp, metrics_v3.box.mr
F13 = 2 * P3 * R3 / (P3 + R3)
print(f"[V3 TEST] mAP50={metrics_v3.box.map50:.4f}, P={P3:.4f}, R={R3:.4f}, F1={F13:.4f}")
show_training_results("solar_panel_v3")
```

```python
# ============================================================
#  Compare Results (V1 vs V2 vs V3)
# ============================================================
print("\n=== TEST Comparison ===")
print(f"{'Model':<10}{'Epochs':<8}{'lr0':<10}{'mAP50':<10}{'Precision':<10}{'Recall':<10}{'F1':<10}")
print(f"{'v1':<10}{'10':<8}{'0.0005':<10}{metrics_v1.box.map50:<10.4f}{P1:<10.4f}{R1:<10.4f}{F11:<10.4f}")
print(f"{'v2':<10}{'20':<8}{'0.001':<10}{metrics_v2.box.map50:<10.4f}{P2:<10.4f}{R2:<10.4f}{F12:<10.4f}")
print(f"{'v3':<10}{'30':<8}{'0.001':<10}{metrics_v3.box.map50:<10.4f}{P3:<10.4f}{R3:<10.4f}{F13:<10.4f}")
```

```python
from ultralytics import YOLO
from PIL import Image
import matplotlib.pyplot as plt
import os

# Define 3 images you want to compare
compare_images = [
    "/content/drive/MyDrive/solar_panel/test/images/image2.jpg",
    "/content/drive/MyDrive/solar_panel/test/images/image4.jpg",
    "/content/drive/MyDrive/solar_panel/test/images/image18.jpg"
]

# Define your models and names
models = {
    "v1 (10 epochs)": "/content/drive/MyDrive/solar_panel_runs/solar_panel_v1/weights/best.pt",
    "v2 (20 epochs)": "/content/drive/MyDrive/solar_panel_runs/solar_panel_v2/weights/best.pt",
    "v3 (30 epochs + aug)": "/content/drive/MyDrive/solar_panel_runs/solar_panel_v3/weights/best.pt"
}

# Run prediction for each model
for model_name, model_path in models.items():
    model = YOLO(model_path)
    results = model.predict(source=compare_images, save=True, name=f"predict_{model_name.replace(' ', '_')}")
    print(f"✅ Predictions done for {model_name}")

# Visualization
fig, axes = plt.subplots(len(compare_images), len(models), figsize=(15, 10))
fig.suptitle("Model Comparison Results (v1, v2, v3)", fontsize=16, y=0.92)

for col, (model_name, _) in enumerate(models.items()):
    pred_path = f"/content/runs/detect/predict_{model_name.replace(' ', '_')}"
    pred_files = sorted([f for f in os.listdir(pred_path) if f.endswith(('.jpg', '.jpeg', '.png'))])

    for row, file in enumerate(pred_files[:len(compare_images)]):
        img = Image.open(os.path.join(pred_path, file))
        axes[row, col].imshow(img)
        axes[row, col].axis("off")
        if row == 0:
            axes[row, col].set_title(model_name, fontsize=12)
        if col == 0:
            axes[row, col].set_ylabel(os.path.splitext(file)[0], fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Helper Function: Predict and visualize random images
# ============================================================
def predict_and_show(model_path, name, n=15):
    model = YOLO(model_path)
    test_dir = "/content/drive/MyDrive/solar_panel/test/images"
    all_imgs = [f for f in os.listdir(test_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
    sample_imgs = random.sample(all_imgs, min(n, len(all_imgs)))

    print(f"Chosen images ({len(sample_imgs)}):", sample_imgs)
    img_paths = [os.path.join(test_dir, img) for img in sample_imgs]

    results = model.predict(source=img_paths, save=True, name=name)
    pred_path = f"/content/runs/detect/{name}"

    # show predictions
    result_imgs = sorted([f for f in os.listdir(pred_path) if f.endswith(('.jpg', '.jpeg', '.png'))])
    cols = 5
    rows = (len(result_imgs) + cols - 1) // cols
    plt.figure(figsize=(20, 4 * rows))
    for i, filename in enumerate(result_imgs):
        img = Image.open(f"{pred_path}/{filename}")
        plt.subplot(rows, cols, i + 1)
        plt.imshow(img)
        plt.axis('off')
        plt.title(filename.split('.')[0], fontsize=9)
    plt.tight_layout()
    plt.show()

```

```python
# ============================================================
# Show random predictions for each model
# ============================================================
predict_and_show("/content/drive/MyDrive/solar_panel_runs/solar_panel_v1/weights/best.pt", "predict_v1_random15")
predict_and_show("/content/drive/MyDrive/solar_panel_runs/solar_panel_v2/weights/best.pt", "predict_v2_random15")
predict_and_show("/content/drive/MyDrive/solar_panel_runs/solar_panel_v3/weights/best.pt", "predict_v3_random15")
```
