```python
from ultralytics import YOLO
import matplotlib.pyplot as plt
from PIL import Image
import os
```

```python
# dataset sturcutre （based on Eri's local path）
data_yaml = "/Users/eri/Desktop/data/data.yaml"
train_path = "/Users/eri/Desktop/data/images/train"
val_path = "/Users/eri/Desktop/data/images/val"
```

```python
# 事前学習済みモデル（COCOデータセットベース）
model = YOLO("yolov8s.pt")
```

```python
model.info(verbose=True)
```

```python
model.train(
    data=data_yaml,     # data setting file
    epochs=50,          # 学習エポック
    imgsz=640,          # 入力画像サイズ
    batch=8,            # バッチサイズ
    lr0=0.0005,         # 初期学習率
    name="train_v1"     # 出力フォルダ名
)
```

```python
metrics = model.val()

print("mAP50-95:", metrics.box.map)
print("mAP50:", metrics.box.map50)
print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
```
