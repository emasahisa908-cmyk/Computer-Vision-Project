```python
!pip3 install ultralytics
from ultralytics import YOLO
from google.colab import drive
import os, random
from PIL import Image
import matplotlib.pyplot as plt

drive.mount('/content/drive')
```

```python
model_v1 = YOLO('yolov8s.pt')
model_v1.train(
    data="/content/drive/MyDrive/solar_panel/data.yaml",
    epochs=10,
    imgsz=640,
    batch=8,
    lr0=0.0005,
    name="solar_panel_v1"
)

metrics_v1_test = model_v1.val(
    data="/content/drive/MyDrive/solar_panel/data.yaml",
    split="test"
)

P1, R1 = metrics_v1_test.box.mp, metrics_v1_test.box.mr
F11 = 2 * P1 * R1 / (P1 + R1)
print(f"[V1 TEST] mAP50={metrics_v1_test.box.map50:.4f}, P={P1:.4f}, R={R1:.4f}, F1={F11:.4f}")
```

```python
model_v2 = YOLO('yolov8s.pt')
model_v2.train(
    data="/content/drive/MyDrive/solar_panel/data.yaml",
    epochs=20,
    imgsz=640,
    batch=8,
    lr0=0.001,
    optimizer='adam',
    name="solar_panel_v2"
)

metrics_v2_test = model_v2.val(
    data="/content/drive/MyDrive/solar_panel/data.yaml",
    split="test"
)

P2, R2 = metrics_v2_test.box.mp, metrics_v2_test.box.mr
F12 = 2 * P2 * R2 / (P2 + R2)
print(f"[V2 TEST] mAP50={metrics_v2_test.box.map50:.4f}, P={P2:.4f}, R={R2:.4f}, F1={F12:.4f}")
```

```python
print("\n=== TEST Comparison (V1 vs V2) ===")
print(f"{'Model':<10}{'lr0':<10}{'Epochs':<8}{'mAP50':<10}{'Precision':<10}{'Recall':<10}{'F1':<10}")
print(f"{'v1':<10}{'0.0005':<10}{'10':<8}{metrics_v1_test.box.map50:<10.4f}{P1:<10.4f}{R1:<10.4f}{F11:<10.4f}")
print(f"{'v2':<10}{'0.001':<10}{'20':<8}{metrics_v2_test.box.map50:<10.4f}{P2:<10.4f}{R2:<10.4f}{F12:<10.4f}")
```

```python
from ultralytics import YOLO
from PIL import Image
import matplotlib.pyplot as plt
import os, random

model = YOLO("/content/runs/detect/solar_panel_v1/weights/best.pt")

# randomly choose images
test_dir = "/content/drive/MyDrive/solar_panel/test/images"
all_imgs = [f for f in os.listdir(test_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]


sample_imgs = random.sample(all_imgs, min(15, len(all_imgs)))
print("chosen images:", sample_imgs)


image_paths = [os.path.join(test_dir, img) for img in sample_imgs]
results = model.predict(source=image_paths, save=True, name="predict_random15_v1")


pred_path = "/content/runs/detect/predict_random15_v1"
print("\n folder:", pred_path)
!ls $pred_path

result_imgs = sorted([f for f in os.listdir(pred_path) if f.endswith(('.jpg', '.jpeg', '.png'))])


cols = 5
rows = (len(result_imgs) + cols - 1) // cols

plt.figure(figsize=(20, 4 * rows))
for i, filename in enumerate(result_imgs):
    img = Image.open(f"{pred_path}/{filename}")
    plt.subplot(rows, cols, i + 1)
    plt.imshow(img)
    plt.axis('off')
    plt.title(filename.split('.')[0], fontsize=10)
plt.tight_layout()
plt.show()
```
